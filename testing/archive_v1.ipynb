#### Importing PDF Files

```python
## Define PDF Loader Function
def load_pdf_file(data):
    """
    Loads all PDF files from the specified directory using LangChain's DirectoryLoader and PyPDFLoader.
    Args:
        data (str): Path to the directory containing PDF files.
    Returns:
        list: A list of loaded document objects.
    """
    loader = DirectoryLoader(data, glob="*.pdf", loader_cls=PyPDFLoader)
    documents = loader.load()
    print(f"Loaded {len(documents)} PDF documents from {data}")
    return documents

## Code block summary: Checks for existing Pinecone index and creates one if needed, then retrieves the index object.
```

```python
## PDF Loader Function
def print_pdf_filenames(data_path):
    """
    Prints the names of all PDF files in the specified directory.
    Args:
        data_path (str): Path to the directory containing PDF files.
    """
    pdf_files = [f for f in os.listdir(data_path) if f.lower().endswith('.pdf')]
    if pdf_files:
        print(f"PDF files in '{data_path}':")
        for fname in pdf_files:
            print(f" - {fname}")
    else:
        print(f"No PDF files found in '{data_path}'.")

## Code block summary: PDF Loader Function - Defines a function to load all PDF documents from a specified directory.
```

```python
## Text Split Function - Define Chunking Function
def text_split(extracted_data):
    text_splitter = RecursiveCharacterTextSplitter(chunk_size=500, chunk_overlap=20)
    text_chunks = text_splitter.split_documents(extracted_data)
    print(f"Split into {len(text_chunks)} chunks")
    return text_chunks

## Code block summary: Defines a function to split loaded documents into smaller text chunks.
```

## Text Chunking Execution

Documents are split into smaller chunks (500 characters with 20-character overlap) to improve retrieval accuracy. This overlap ensures that context isn't lost at chunk boundaries, which is critical for maintaining semantic coherence.

```python
## Hugging Face Function - Download the Embeddings from Hugging Face
def download_hugging_face_embeddings():
    embeddings = HuggingFaceEmbeddings(
        model_name="sentence-transformers/all-MiniLM-L6-v2",
        model_kwargs={'device': 'cpu'},
        encode_kwargs={'normalize_embeddings': True}
    )
    return embeddings

embeddings = download_hugging_face_embeddings()

## Code block summary: Downloads and initializes Hugging Face sentence transformer embeddings.
```

## Embedding Model Initialization

We use Hugging Face's `all-MiniLM-L6-v2` model to generate embeddings. This model:
- Creates 384-dimensional vector representations of text
- Runs efficiently on CPU
- Normalizes embeddings for better cosine similarity calculations
- Balances performance and quality for semantic search tasks

```python
## Define a function to upsert documents to Pinecone
def upsert_to_pinecone(documents, embedding_function, batch_size=100):
    """
    Upsert documents to Pinecone index.
    Args:
        documents: List of LangChain Document objects
        embedding_function: Function to generate embeddings
        batch_size: Size of batches for upserting
    """
    for i in range(0, len(documents), batch_size):
        batch = documents[i:i+batch_size]
        # Create IDs and embed documents
        ids = [f'doc_{i+j}' for j in range(len(batch))]
        texts = [doc.page_content for doc in batch]
        metadatas = [
            {
                'text': doc.page_content,
                **doc.metadata
            } 
            for doc in batch
        ]
        # Get embeddings
        embeddings_batch = embedding_function.embed_documents(texts)
        # Prepare vectors for upserting
        vectors = [
            {
                'id': ids[j],
                'values': embeddings_batch[j],
                'metadata': metadatas[j]
            }
            for j in range(len(batch))
        ]
        # Upsert to Pinecone
        index.upsert(vectors=vectors)
    print(f"Upserted {len(documents)} documents to Pinecone index '{index_name}'")

## Code block summary: Defines a function to upsert document embeddings into the Pinecone index in batches.
```

## Uploading Embeddings to Pinecone

This cell executes the upsert operation, sending all document chunks (with their embeddings and metadata) to Pinecone. The batch processing approach (100 documents at a time) ensures efficient uploads without overwhelming the API.

```python
class PineconeRetriever(BaseRetriever):
    k: int = 3

    def __init__(self, index, embedding_function, k=3, filter=None, namespace=None, **kwargs):
        super().__init__(k=k, **kwargs)
        self._index = index
        self._embedding_function = embedding_function
        self._filter = filter
        self._namespace = namespace

    def _get_relevant_documents(
        self, query: str, *, run_manager: CallbackManagerForRetrieverRun
    ) -> List[Document]:
        query_embedding = self._embedding_function.embed_query(query)
        results = self._index.query(
            vector=query_embedding,
            top_k=self.k,
            include_metadata=True,
            filter=self._filter,
            namespace=self._namespace
        )
        documents = []
        for match in results["matches"]:
            metadata = match.get("metadata", {})
            doc = Document(
                page_content=metadata.get("text", ""),
                metadata={k: v for k, v in metadata.items() if k != "text"}
            )
            documents.append(doc)
        return documents
```

## Custom Pinecone Retriever

This custom retriever class integrates Pinecone with LangChain's retrieval framework. It:
- Converts user queries into embeddings
- Searches Pinecone for the top-k most similar documents
- Returns results as LangChain Document objects
- Supports filtering and namespace operations for advanced use cases

```python
# Create a retriever
retriever = PineconeRetriever(
    index=index,
    embedding_function=embeddings,
    k=3
)

# Test the retriever
retrieved_docs = retriever.invoke("Who is Benjamin Graham?")
print(retrieved_docs)
```

```python
## Set up Language Model (LLM) for question-answering
# Configure ChatOpenAI with specific parameters:
# - temperature: Controls randomness of output (0.7 allows some creativity)
# - max_tokens: Limits the response length
# - model: Specifies the OpenAI model to use (GPT-4o-mini in this case)
# - api_key: Securely loaded from environment variables
# - seed: Ensures reproducibility of generated responses
llm = ChatOpenAI(
    temperature=0.7,
    max_tokens=500,
    #model="gpt-3.5-turbo", ## Commented out for testing purposes
    model="gpt-4o-mini",
    api_key=OPENAI_API_KEY,
    seed = 42
)

## print model type 
print(f"Using model: {llm.model_name}")

## Define the system prompt for the question-answering task
# This system prompt provides context and guidelines for the AI's responses
# Key aspects:
# - Specifies the AI's role as a personal finance expert
# - Instructs to base answers strictly on provided context
# - Provides guidance on handling insufficient or conflicting information
# - Sets expectations for answer length and style
system_prompt = (
    "You are a helpful, knowledgeable assistant for question-answering tasks related to personal finance. You are a personal finance expert. Your audience may be knowledgeable or not, so you should explain concepts clearly and concisely. "
    "Base your answers strictly on the retrieved context provided below. "
    "If the context does not contain enough information to answer the question, reply ONLY with 'I don't know.' "
    "If the context contains conflicting information, mention this briefly. "
    "Keep your answer concise (maximum three sentences), clear, and informative. "
    "Cite or reference the context when possible. "
    "Do not use any outside knowledge or make assumptions."
    "\n\n"
    "{context}"
)

# Create a prompt template that combines the system prompt with user input
prompt = ChatPromptTemplate.from_messages([
    ("system", system_prompt),
    ("human", "{input}"),
])

## Create the question-answering chain
# This involves two main components:
# 1. create_stuff_documents_chain: Combines the LLM with the prompt
# 2. create_retrieval_chain: Adds document retrieval capability
question_answer_chain = create_stuff_documents_chain(llm, prompt)
rag_chain = create_retrieval_chain(retriever, question_answer_chain)
```

## RAG Chain Construction

This is the heart of our system where we assemble all components into a complete RAG pipeline:

1. **Language Model Setup**: Configure GPT-4o-mini with specific parameters
   - Temperature: 0.7 for balanced creativity and factuality
   - Max tokens: 500 to keep responses concise
   - Seed: 42 for reproducibility
   
2. **System Prompt Design**: Craft instructions that ensure the model:
   - Acts as a personal finance expert
   - Bases answers strictly on retrieved context
   - Admits when it doesn't know something
   - Keeps responses concise and cites sources
   
3. **Chain Assembly**: Combine the retriever and LLM into a cohesive RAG chain that:
